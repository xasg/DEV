## Homologar nombres exactos

### Subtask:
Realizar una unión (merge) inicial entre `df_instrumento` y `df_anuies` utilizando los nombres de las instituciones como clave. Esto identificará las coincidencias exactas.

## Identificar nombres sin coincidencia exacta

### Subtask:
Determinar qué nombres de instituciones en `df_instrumento` no tuvieron una coincidencia exacta en `df_anuies`.


**Reasoning**:
Perform a left merge of df_instrumento with df_instrumento_homologado on 'ID_unico' to identify the rows that did not have an exact match in df_anuies.



```python
df_merged = pd.merge(
    df_instrumento,
    df_instrumento_homologado[['ID_unico', 'INSTITUCIÓN']],
    on='ID_unico',
    how='left',
    suffixes=('_instrumento', '_homologado')
)

instituciones_no_coincidentes = df_merged[df_merged['INSTITUCIÓN'].isnull()]['Nombre de la Institución de Educación Superior: ']

display(instituciones_no_coincidentes.head())
```

## Homologación aproximada (con ia si es necesario)

### Subtask:
For the institution names without an exact match, explore techniques for approximate string matching (fuzzy matching). If complexity requires it, consider using AI models to find the best possible matches.


**Reasoning**:
Now that fuzzywuzzy is installed, import the fuzz module and initialize the potential matches dictionary.



```python
from fuzzywuzzy import fuzz

potential_matches = {}
```

**Reasoning**:
Iterate through the non-matching institutions, calculate similarity scores with ANUIES institutions, find the best match above a threshold, and store potential matches for review.



```python
threshold = 80
potential_matches = {}

# Combine institution name and acronym for better matching
df_anuies['INSTITUCION_SIGLAS'] = df_anuies['INSTITUCIÓN'] + ' (' + df_anuies['SIGLAS'] + ')'

for original_name in instituciones_no_coincidentes:
    best_match = None
    highest_score = 0
    for index, row in df_anuies.iterrows():
        # Calculate score with institution name
        score_name = fuzz.token_sort_ratio(original_name, row['INSTITUCIÓN'])
        # Calculate score with institution name and acronym
        score_siglas = fuzz.token_sort_ratio(original_name, row['INSTITUCION_SIGLAS'])

        score = max(score_name, score_siglas)

        if score > highest_score:
            highest_score = score
            best_match = row['INSTITUCIÓN'] # Store the full institution name from df_anuies
    if highest_score >= threshold:
        potential_matches[original_name] = {'match': best_match, 'score': highest_score}

# Print potential matches for manual review
for original, match_info in potential_matches.items():
    print(f"Original: {original} -> Potential Match: {match_info['match']} (Score: {match_info['score']})")
```

### Identificar y seleccionar registros con nombres homologados repetidos

**Subtask:** Agrupar el DataFrame por la columna `Nombre_Homologado`, identificar los grupos con más de un registro (nombres repetidos) y, para cada grupo, seleccionar la fila con la menor cantidad de valores nulos para incluir en un nuevo DataFrame.

```python
instituciones_sin_homologar = df_instrumento[df_instrumento['Nombre_Homologado'].isnull()]
display(instituciones_sin_homologar[['Nombre de la Institución de Educación Superior: ', 'Nombre_Homologado']])
```

```python
# Count the occurrences of each unique value in 'Nombre_Homologado'
conteo_nombres_homologados = df_instrumento['Nombre_Homologado'].value_counts()

# Identify the number of values with more than one occurrence
numero_nombres_homologados_repetidos = len(conteo_nombres_homologados[conteo_nombres_homologados > 1])

# Display the count
print("Número de nombres homologados con más de una incidencia:", numero_nombres_homologados_repetidos)
```

```python
# Identify duplicated 'Nombre_Homologado' values
nombres_homologados_duplicados = df_instrumento[df_instrumento['Nombre_Homologado'].duplicated(keep=False)].copy()

# Calculate the number of nulls for each row
nombres_homologados_duplicados['num_nulls'] = nombres_homologados_duplicados.isnull().sum(axis=1)

# Group by 'Nombre_Homologado' and find the index of the row with the minimum number of nulls in each group
idx_min_nulls = nombres_homologados_duplicados.groupby('Nombre_Homologado')['num_nulls'].idxmin()

# Select the rows with the minimum number of nulls
registros_unicos_por_nombre_homologado = nombres_homologados_duplicados.loc[idx_min_nulls].drop(columns=['num_nulls'])

# Display the new DataFrame
display(registros_unicos_por_nombre_homologado)
registros_unicos_por_nombre_homologado.shape
```

### Crear DataFrame final sin duplicados y comparar

**Subtask:** Construir el DataFrame final sin duplicados en la columna `Nombre_Homologado`, utilizando los registros únicos seleccionados para los nombres duplicados y manteniendo los registros no duplicados, y luego comparar el número de filas con el DataFrame original.

```python
# Get the indices of the rows to keep from the original DataFrame
# These are the indices of the non-duplicated rows and the indices of the unique rows selected from the duplicated ones
indices_to_keep = df_instrumento[~df_instrumento['Nombre_Homologado'].duplicated(keep=False)].index.tolist() + registros_unicos_por_nombre_homologado.index.tolist()

# Create the new DataFrame by selecting the rows to keep from the original DataFrame
df_instrumento_sin_duplicados = df_instrumento.loc[indices_to_keep].reset_index(drop=True)


# Get the number of rows in the original and new DataFrames
num_filas_original = df_instrumento.shape[0]
num_filas_sin_duplicados = df_instrumento_sin_duplicados.shape[0]

# Calculate the number of duplicates removed
registros_duplicados_eliminados = num_filas_original - num_filas_sin_duplicados

# Merge df_instrumento_sin_duplicados with df_anuies to get the 'ESTADO'
df_instrumento_sin_duplicados = pd.merge(
    df_instrumento_sin_duplicados,
    df_anuies[['INSTITUCIÓN', 'ESTADO']],
    left_on='Nombre_Homologado',
    right_on='INSTITUCIÓN',
    how='left'
)

# Drop the extra 'INSTITUCIÓN' column from the merge
df_instrumento_sin_duplicados = df_instrumento_sin_duplicados.drop(columns=['INSTITUCIÓN'])

# Display the head of the updated DataFrame to verify
df_instrumento_sin_duplicados.shape

# Display the comparison
print(f"Número de filas en el DataFrame original ({df_instrumento}): {num_filas_original}")
print(f"Número de filas en el nuevo DataFrame (sin duplicados en Nombre_Homologado): {num_filas_sin_duplicados}")
print(f"Número de registros duplicados eliminados: {registros_duplicados_eliminados}")

# Save dataframe
df_instrumento_sin_duplicados.to_csv('df_instrumento_sin_duplicados.csv', index=False)

# Display the head of the new DataFrame
display(df_instrumento_sin_duplicados.head(2))
```

### Crear DataFrame con registros eliminados

**Subtask:** Identificar y crear un nuevo DataFrame que contenga solo los registros duplicados que fueron eliminados al generar `df_instrumento_sin_duplicados`.

```python
# Identify all duplicated rows in the original DataFrame based on 'Nombre_Homologado'
all_duplicated_rows = df_instrumento[df_instrumento['Nombre_Homologado'].duplicated(keep=False)]

# Identify the indices of the rows that were kept for duplicated names
indices_of_kept_duplicates = registros_unicos_por_nombre_homologado.index

# Filter the duplicated rows to exclude the ones that were kept
df_registros_eliminados = all_duplicated_rows.drop(indices_of_kept_duplicates)

# Display the new DataFrame with eliminated records
#display(df_registros_eliminados.head())

df_registros_eliminados.to_csv('registros_eliminados.csv', index=False)


# Display the number of eliminated records to verify
print(f"Número de registros eliminados: {df_registros_eliminados.shape[0]}")
```

```python
# Merge df_instrumento_sin_duplicados with df_anuies to get the 'ESTADO'
df_instrumento_sin_duplicados = pd.merge(
    df_instrumento_sin_duplicados,
    df_anuies[['INSTITUCIÓN', 'ESTADO']],
    left_on='Nombre_Homologado',
    right_on='INSTITUCIÓN',
    how='left'
)

# Drop the extra 'INSTITUCIÓN' column from the merge
df_instrumento_sin_duplicados = df_instrumento_sin_duplicados.drop(columns=['INSTITUCIÓN'])

# Display the head of the updated DataFrame to verify
display(df_instrumento_sin_duplicados.head(2))
```
